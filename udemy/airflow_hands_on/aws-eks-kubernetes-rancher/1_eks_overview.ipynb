## AWS Elastic Kubernetes Service (EKS) Overview

AWS EKS
- makes it easy to deploy, manage, and scale containerized apps
- orchestrate your containers
- still need to configure subnets, vps, iam roles, ssh keys
- integrated with aws services (ECR for container images, Elastic Load Balancing - ELB, IAM, etc.)

## Create a new EC2 Instance for Rancher

Go to AWS Console and create a new instance (Amazon Linux 2023) for rancher with at least 2 gbs (t2.small)

Connect to the new instance and run:
```bash
sudo yum update -y

sudo yum install docker

# start docker service
sudo service docker start

# add ec2-user to docker group to execute docker commands w/o sudo
sudo usermod -a -G docker ec2-user

# close connection then connect again to apply new docker permissions

# run rancher in docker
docker run -d \
--restart=unless-stopped \
--name rancher \
--hostname rancher \
-p 80:80 \
-p 443:443 \
rancher/rancher:v2.3.2

# visit the racher url and use admin/admin for login
```

## Configure ECR

name: airflow

Install AWS CLI in your machine and configure airflow-user

Follow push commands for airflow repo in ECR
```bash
# Retrieve an authentication token and authenticate your Docker client to your registry
aws ecr get-login-password --region ap-southeast-2 --profile airflowuser | docker login --username AWS --password-stdin 222634372385.dkr.ecr.ap-southeast-2.amazonaws.com

# Build your Docker image
docker build -t airflow .

# tag your image so you can push the image to this repository
docker tag airflow:latest 222634372385.dkr.ecr.ap-southeast-2.amazonaws.com/airflow:latest

# push this image to your newly created AWS repository
docker push 222634372385.dkr.ecr.ap-southeast-2.amazonaws.com/airflow:latest

```

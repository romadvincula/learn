```pyspark
# read processed files by emr into a dataframe
df = spark.read \
    .parquet('s3://romadv-itv-github-emr/prod/raw/ghactivity/')
```

```pyspark
df.printSchema()
```

```pyspark
df.show(5)
```

```pyspark
df.count()
```

```pyspark
df.select('created_at').show(5, truncate=False)
```

```pyspark
df.filter("substr(created_at, 1, 10) == '2021-01-13'").count()
```

```pyspark
from pyspark.sql.functions import expr, count, lit
```

```pyspark
df.groupBy(expr('substr(created_at, 1, 10)').alias('created_date')) \
    .agg(count(lit("*")).alias("activity_count")) \
    .show()
```

We need to make sure that the user has required permissions to manage Glue Catalog via CLI.

We can assign Glue Service Role to the user whose credentials are configured as part of AWS CLI Profile.

```python
import boto3
import os

os.environ.setdefault('AWS_PROFILE', 'itvgithub')
os.environ.setdefault('AWS_DEFAULT_REGION', 'ap-southeast-2')
```

```python
!echo $AWS_PROFILE
```

```python
glue_client = boto3.client('glue', verify=False)
```

```python
# list crawlers
glue_client.list_crawlers()
```

```python
# get name of a specific crawler
crawler = glue_client.list_crawlers()['CrawlerNames'][3]
```

```python
# get crawler
glue_client.get_crawler(
    Name=crawler
)
```

```python
# trigger crawler run
glue_client.start_crawler(
    Name=crawler
)
```

```python
# get crawler state
glue_client.get_crawler(
    Name=crawler
)['Crawler']['State']
```

```python
# get details about databases
for db in glue_client.get_databases()['DatabaseList']:
    print(db['Name'])
```

```python
# get details about tables and for a specific table
for table in glue_client.get_tables(
    DatabaseName='retail_db'
)['TableList']:
    print(table['Name'])
 
glue_client.get_table(
    DatabaseName='retail_db',
    Name='orders'
)
```

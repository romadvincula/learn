```sql
-- Get list of tables querying information schema
SELECT * FROM information_schema.tables
WHERE table_schema = 'public';

-- Validate users data using Query Editor
SELECT * FROM users LIMIT 10;
 
SELECT count(*) FROM users;

-- Create Redshift Table
CREATE TABLE myusers (
  user_id INT PRIMARY KEY,
  user_first_name VARCHAR(30),
  user_last_name VARCHAR(30)
);

SELECT COUNT(*) FROM myusers;


-- Insert data into table
INSERT INTO myusers
(user_id, user_first_name, user_last_name)
VALUES
(1, 'Scott', 'Tiger');

INSERT INTO myusers
VALUES
(2, 'Donald', 'Duck');

-- primary key uniqueness is not enforced (and also foreign keys) so these will stil be inserted
INSERT INTO myusers
VALUES
(1, 'Scott', 'Tiger'),
(2, 'Donald', 'Duck');

-- not null constraint is enforced so this will fail
INSERT INTO myusers
(user_first_name, user_last_name)
VALUES
('Donald', 'Duck');
-- Note: index is also not supported in redshift

SELECT * FROM myusers;


-- CRUD Operations - Update records in Redshift table
UPDATE myusers
	SET user_first_name = 'Mickey', user_last_name = 'Mouse'
WHERE user_id = 2;
 
SELECT * FROM myusers;
 
SELECT lower('Mouse');
 
SELECT * FROM myusers;
 
UPDATE myusers
	SET user_first_name = lower(user_first_name), user_last_name = lower(user_last_name);
```

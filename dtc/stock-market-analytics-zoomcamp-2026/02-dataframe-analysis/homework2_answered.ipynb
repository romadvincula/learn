## Question 1: [IPO] Withdrawn IPOs by Company Type

### What is the total withdrawn IPO value (in $ millions) for the company class with the highest total withdrawal value?

```python
import pandas as pd
import re
import numpy as np
```

```python
url = "https://www.iposcoop.com/ipos-recently-filed/"
ipos_recently_filed = pd.read_html(url)[0]

# look at the ipo data
ipos_recently_filed.head(3)
```

```python
ipos_recently_filed.info()
```

```python
# filter to 'Withdrawn' only for Expected to Trade
withdrawn_df = ipos_recently_filed[ipos_recently_filed['Expected To Trade']=='Withdrawn']
display(withdrawn_df.shape)
```

```python
# classifies company type based on keywords
def classify_company_type(company_name):
    company_name_lower = str(company_name).lower()
    if 'technologies' in company_name_lower:
        return 'Technologies'
    elif 'acquisition corp' in company_name_lower or \
         'acquisition corporation' in company_name_lower or \
         re.search(r'\bcorp\b', company_name_lower):
        return 'Acquisition Corp'
    elif 'inc' in company_name_lower or 'incorporated' in company_name_lower:
        return 'Inc.'
    elif 'group' in company_name_lower:
        return 'Group'
    elif 'ltd' in company_name_lower or 'limited' in company_name_lower:
        return 'Limited'
    elif 'holdings' in company_name_lower or 'holding' in company_name_lower:
        return 'Holdings'
    else:
        return 'Other'

withdrawn_df['Company Type'] = withdrawn_df['Company'].apply(classify_company_type)
withdrawn_df.head(3)
```

```python
def find_avg_price(row: pd.DataFrame):
  low = pd.to_numeric(str(row["Price Low"]).replace('$', ''), errors="coerce")
  high = pd.to_numeric(str(row["Price High"]).replace('$', ''), errors="coerce")

  return (low + high) / 2

withdrawn_df['Avg_price'] = withdrawn_df.apply(find_avg_price, axis=1)
withdrawn_df.head(3)
```

```python
withdrawn_df["Shares (millions)"] = pd.to_numeric(withdrawn_df['Shares (millions)'], errors="coerce")
withdrawn_df['Est $ Vol (millions)'] = pd.to_numeric(withdrawn_df['Est $ Vol (millions)'].str.replace('$', '').replace(',', ''), errors="coerce")
withdrawn_df.head(3)
```

```python
withdrawn_df['Shares_offered_value'] = np.where(
    pd.isna(withdrawn_df['Shares (millions)'] * withdrawn_df['Avg_price']),
    withdrawn_df['Est $ Vol (millions)'],
    withdrawn_df['Shares (millions)'] * withdrawn_df['Avg_price']
)
withdrawn_df.head()
```

```python
agg_df = withdrawn_df.groupby(['Company Type']) \
.Shares_offered_value.agg(['sum']) \
.rename({'sum': 'sum_shares_offered_value'}, axis=1)
agg_df
```

```python
agg_df.sort_values(by='sum_shares_offered_value', ascending=False)
```

### Answer: The class 'Acquisition Corp' has the highest total value of withdrawals (outside of Other) with $499.9850.

## Question 2: [IPO] Median Sharpe Ratio for 2025 IPOs (First 8 Months)

### What is the median Sharpe ratio (as of 11 September 2026) for companies that went public before 1 September 2025?

```python
import pandas as pd
from time import sleep
import yfinance as yf
import numpy as np
import datetime as dt
```

```python
url = "https://www.iposcoop.com/2025-pricings/"
ipos25 = pd.read_html(url)[0]
display(ipos25.head(3))
display(ipos25.info())

# convert datatypes to correct format
ipos25["Offer Date"] = pd.to_datetime(ipos25["Offer Date"])
ipos25["Return"] = pd.to_numeric(ipos25["Return"].str.replace("%", ''), errors="coerce")
```

```python
# filter to only before sep 1 2025 offer date and remove those with 0% return
filtered_df = ipos25[(ipos25["Offer Date"] < "2025-09-01") & (ipos25["Return"] != 0)]
# filtered_df = ipos25[ipos25["Offer Date"] < "2025-09-01"]
# filtered_df = filtered_df[filtered_df["Return"] != 0.0]
display(filtered_df.head(3))
display(filtered_df.shape)
```

```python
# get daily stocks prices
tickers = filtered_df['Symbol'].unique()
stocks_df = pd.DataFrame({'A': []})
start = dt.date(year=2024, month=1, day=1)
end = dt.date(year=2026, month=9, day=15)
risk_free_rate_of_return = 0.05
delisted = []

for i, t in enumerate(tickers):
  # print(i, t)
  try:
    # get data for the current ticker
    temp = yf.Ticker(t).history(start=start, end=end, interval="1d")

    # transformations
    temp['ticker'] = t
    temp['Date'] = pd.to_datetime(temp.index.date)
    temp['year'] = temp.index.year
    temp['month'] = temp.index.month
    temp['weekday'] = temp.index.weekday
    temp['close_252d'] = temp.Close.shift(252)
    # 1 year of trading day's growth
    temp['growth_252d'] = temp.Close / temp['close_252d']
    temp['growth_252d_percent'] = (temp.Close / temp['close_252d']) - 1
    # annualized volatility
    temp['volatility'] = temp.Close.rolling(30).std() * np.sqrt(252)
    # sharpe ratio - risk-adjusted performance of an investment (higher is better)
    temp['sharpe'] = (
        temp['growth_252d'] - risk_free_rate_of_return
        ) / temp['volatility']

    # calculate future growth for question 3
    for month_num in range(1, 13, 1):
      temp["future_growth_"+str(month_num)+"_m"] = (
          temp.Close.shift(-21 * month_num) / temp.Close
      )


  except Exception as e:
    print(f"Skipped {t} due to {e}")
    delisted.append(t)
    continue

  else:
    # save into main dataframe
    if stocks_df.empty:
      stocks_df = temp
    else:
      stocks_df = pd.concat([stocks_df, temp], ignore_index=True)

  sleep(1)
```

```python
stocks_df.tail(3)
```

```python
stocks_filtered = stocks_df[stocks_df.Date == '2026-09-11']
display(stocks_filtered.sharpe.describe())
```

### Answer: The median Sharpe ratio is 0.050142 or 0.04.

## Question 3: [IPO] 'Fixed Months Holding Strategy'

### What is the optimal number of months (1 to 12) to hold a newly IPO'd stock in order to maximize the median growth value?

```python
display(stocks_df.head(3))
display(stocks_df.info())
```

```python
display(filtered_df.tail(3))
display(filtered_df.info())
```

```python
# convert 1st day close to numeric
filtered_df["close_1st_day"] = pd.to_numeric(filtered_df["1st Day Close"].str.replace("$", ""), errors="coerce")
entry_df = filtered_df[["Symbol", "Offer Date", "close_1st_day"]]
entry_df.rename(columns={"Symbol": "ticker", "Offer Date": "min_date"}, inplace=True)
entry_df["ticker-mindate"] = entry_df["ticker"] + entry_df["min_date"].dt.strftime("%Y-%m-%d")
entry_df.head(3)
```

```python
# create join column for later
stocks_df["ticker-date"] = stocks_df["ticker"] + stocks_df["Date"].dt.strftime("%Y-%m-%d")
stocks_df.head(3)
```

```python
merged_df = pd.merge(
    entry_df,
    stocks_df,
    how="inner",
    left_on="ticker-mindate",
    right_on="ticker-date",
    validate="many_to_one"
)[["ticker_x", "min_date", "Close", "close_1st_day",
   "future_growth_1_m", "future_growth_2_m", "future_growth_3_m", "future_growth_4_m",
   "future_growth_5_m", "future_growth_6_m", "future_growth_7_m", "future_growth_8_m",
   "future_growth_9_m", "future_growth_10_m", "future_growth_11_m", "future_growth_12_m"
   ]]
display(merged_df.head())
```

```python
merged_df.describe()
```

### Answer: The optimal holding period is only 1 month with median future growth of 0.922104. The value of the newly IPO'd stocks tend to decrease in its first 12 months after the IPO.

## Question 4: [Strategy] Simple RSI-Based Trading Strategy

### What is the total profit (in $ thousands) you would have earned by investing $1000 every time a stock was oversold (RSI < 30)?

```python
import gdown
import pandas as pd
file_id = "1grCTCzMZKY5sJRtdbLVCXg8JXA8VPyg-"
gdown.download(f"https://drive.google.com/uc?id={file_id}", "data.parquet", quiet=False)
df = pd.read_parquet("data.parquet", engine="pyarrow")
```

```python
display(df.head(3))
display(df.info())
```

```python
oversold_signal = 30

# filter date range
filtered = df[(df["Date"] > "2000-01-01") & (df["Date"] < "2025-06-01")]
# filter on buy signal
filtered = filtered[filtered["rsi"] < oversold_signal]
# trim down columns
filtered = filtered[[
    "Ticker", "Date", "Close_x", "rsi", "growth_future_30d"
]]
# calculate income
filtered["income"] = 1000 * (filtered["growth_future_30d"] - 1)
filtered["profitable"] = filtered["income"] > 0
display(filtered.shape)
display(filtered.head())
display((filtered["profitable"] == True).sum() / filtered.shape[0])
```

```python
filtered.describe()
```

```python
net_income = filtered["income"].sum()
display(net_income)
```

### Answer: The net income is about $65,805.
